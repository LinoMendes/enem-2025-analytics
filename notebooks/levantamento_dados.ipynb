```python
from pathlib import Path
import pandas as pd

# Sobe um nível a partir de notebooks/ e aponta para dados/
pasta_dados = Path("../dados")

print("Arquivos encontrados em dados/:\n")
for arquivo in sorted(pasta_dados.rglob("*")):
    if arquivo.is_file():
        tamanho_mb = arquivo.stat().st_size / (1024 * 1024)
        print(f"{arquivo}  -->  {tamanho_mb:.2f} MB")
```

```python
# Localiza automaticamente todos os arquivos .csv dentro de dados/
arquivos_csv = sorted(pasta_dados.rglob("*.csv"))

for caminho_csv in arquivos_csv:
    print("=" * 80)
    print(f"ARQUIVO: {caminho_csv.name}")
    print("=" * 80)
    
    # Lê apenas 5 linhas usando separador ';' e encoding 'latin1' (padrão INEP)
    df_amostra = pd.read_csv(
        caminho_csv,
        sep=";",
        encoding="latin1",
        nrows=5
    )
    
    print(f"Total de colunas: {len(df_amostra.columns)}")
    print("\nLista de colunas e tipos inferidos:")
    print(df_amostra.dtypes)
    
    print("\nPrimeiras 5 linhas:")
    display(df_amostra)
    print("\n")
```

```python
caminho_resultados = next(pasta_dados.rglob("RESULTADOS_2025.csv"))

df_res = pd.read_csv(
    caminho_resultados,
    sep=";",
    encoding="latin1",
    nrows=5
)

for idx, (col, dtype) in enumerate(df_res.dtypes.items(), start=1):
    print(f"{idx:02d}. {col:<25} ({dtype})")
```
